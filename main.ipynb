# 3. Outlier Handling

●	Use Z-score method to detect fare & distance anomalies.

```python
df_b= df_trip.copy()
df_b["fare_amount_z"]=(df_b["fare_amount"]-df_b["fare_amount"].mean())/df_b["fare_amount"].std() 
outliers_fm= df_b[df_b["fare_amount_z"].abs() >3]
```

```python
print(df_b.shape)
print(df_trip.shape)
```

for distance

```python
df_b["distance_z"]=(df_b["distance_km"]-df_b["distance_km"].mean())/df_b["distance_km"].std() 
outliers_d= df_b[df_b["distance_z"].abs() >3]
```

```python
print(df_b.shape)
print(df_trip.shape)
```

●	Use IQR method for ride duration anomalies.

```python
df=df_trip.copy()
q1=df["duration_min"].quantile(0.25)
q3=df["duration_min"].quantile(0.75)
IQR=q3-q1
lower=q1-1.5*IQR
upper=q3-1.5*IQR
```

```python
df["duration_min"]=df["duration_min"].clip(lower=lower,upper=upper)
```

```python
print(df_trip.shape)
print(df.shape)
```

●	Apply Winsorization where necessary (e.g., extreme surge fares).

```python
df1=df_trip.copy()
lower = df1["surge_flag"].quantile(0.01)
upper = df1["surge_flag"].quantile(0.99)
    
df1["surge_flag"] = df1["surge_flag"].clip(
    lower=lower,
    upper=upper)
```

```python
print(df_trip.shape)
print(df1.shape)
```

# 6. Feature Construction

```python
rider_s = df_T.groupby('rider_id').agg(
    total_distance=('distance_km', 'sum'),
    total_fare=('fare_amount', 'sum'),
    total_trips=('trip_id', 'count')
).reset_index()
```

```python
df_r["avg_ride_distance"]=rider_s["total_distance"]/rider_s["total_trips"]
```

```python
df_r["avg_ride_fare"]=rider_s["total_fare"]/rider_s["total_trips"]
```

```python
peak_hours = {7, 8, 9, 18, 19, 20, 21}

if 'ride_time' in df_T.columns:
    trip_hours = pd.to_datetime(df_T['ride_time']).dt.hour
    df_T['is_peak_hour'] = trip_hours.isin(peak_hours).astype(int)
elif 'hour' in df_T.columns:
    df_T['is_peak_hour'] = df_T['hour'].isin(peak_hours).astype(int)
```

```python
df_r['signup_date'] = pd.to_datetime(df_r['signup_date'])
today = pd.to_datetime('today').normalize()
df_r['days_since_signup'] = (today - df_r['signup_date']).dt.days
```

```python
df_r['ride_cancellation_rate'] = (df_r['cancelled_rides'] / df_r['total_rides'].replace(0, np.nan)).fillna(0)
```

```python
fare_per_km = df_T['fare_amount'] / df_T['distance_km']
threshold = 20
df_T['surge_flag'] = (fare_per_km > threshold).astype(int)
```
